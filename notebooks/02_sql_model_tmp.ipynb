```python
import duckdb
con = duckdb.connect(f'{DATA_DIR}/nbi.duckdb')

%load_ext sql
%sql con
%config SqlMagic.displaylimit = 20
```

```python
from pathlib import Path

raw_dir = Path(f'{DATA_DIR}/raw')

for f in sorted(raw_dir.rglob('*.txt')):
    size_mb = f.stat().st_size / 1e6
    print(f, round(size_mb), 'MB')
```

```python
path_2024 = next(raw_dir.rglob('*2024*.txt'))
print(path_2024)

all_cols = con.sql(f"""
    SELECT *
    FROM read_csv('{path_2024}',
                  delim=',',
                  quote='''',
                  header=true,
                  all_varchar=true)
    LIMIT 0
""").columns

print(len(all_cols), 'columns')
print(all_cols)
```
